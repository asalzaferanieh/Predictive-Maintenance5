```python
df = pd.read_csv("ai4i2020.csv")
df = df.drop(columns=["UDI", "Product ID", "TWF", "HDF", "PWF", "OSF", "RNF"])
df = pd.get_dummies(df, columns=["Type"], drop_first=True)

# کلاسیفیکیشن
X_clf = df.drop(columns=["Machine failure"])
y_clf = df["Machine failure"]

# رگرسیون
X_reg = df.drop(columns=["Machine failure", "Torque [Nm]"])
y_reg = df["Torque [Nm]"]
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

Xc_tr, Xc_te, yc_tr, yc_te = train_test_split(
    X_clf, y_clf, test_size=0.2, random_state=42, stratify=y_clf)


Xr_tr, Xr_te, yr_tr, yr_te = train_test_split(
    X_reg, y_reg, test_size=0.2, random_state=42)


sc_c = StandardScaler().fit(Xc_tr)
Xc_tr_s, Xc_te_s = sc_c.transform(Xc_tr), sc_c.transform(Xc_te)

sc_r = StandardScaler().fit(Xr_tr)
Xr_tr_s, Xr_te_s = sc_r.transform(Xr_tr), sc_r.transform(Xr_te)
```

Classification & Tensorflow

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, f1_score, recall_score

lr = LogisticRegression(class_weight="balanced").fit(Xc_tr_s, yc_tr)
rf = RandomForestClassifier(class_weight="balanced", random_state=42).fit(Xc_tr_s, yc_tr)

res_c = {}
for name, m in [("LogReg", lr), ("RandomForest", rf)]:
    p = m.predict(Xc_te_s)
    print(name); print(confusion_matrix(yc_te, p)); print(classification_report(yc_te, p))
    res_c[name] = (recall_score(yc_te, p), f1_score(yc_te, p))
```

```python
model_c = keras.Sequential([
    layers.Input(shape=(Xc_tr_s.shape[1],)),
    layers.Dense(32, activation='relu'),
    layers.Dense(16, activation='relu'),
    layers.Dense(8, activation='relu'),
    layers.Dense(1, activation='sigmoid')
])
model_c.compile(optimizer='adam', loss='binary_crossentropy')
model_c.fit(Xc_tr_s, yc_tr, epochs=50, batch_size=32,
            validation_split=0.2, class_weight={0:1.0, 1:10.0}, verbose=0)

p = (model_c.predict(Xc_te_s) > 0.5).astype(int)
print(confusion_matrix(yc_te, p)); print(classification_report(yc_te, p))
res_c["TensorFlow"] = (recall_score(yc_te, p), f1_score(yc_te, p))
```

Regression & Tensoeflow 

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score

linr = LinearRegression().fit(Xr_tr_s, yr_tr)
rfr = RandomForestRegressor(random_state=42).fit(Xr_tr_s, yr_tr)

res_r = {}
for name, m in [("LinearReg", linr), ("RandomForest", rfr)]:
    p = m.predict(Xr_te_s)
    res_r[name] = (mean_squared_error(yr_te, p) ** 0.5, r2_score(yr_te, p))
```

```python
model_r = keras.Sequential([
    layers.Input(shape=(Xr_tr_s.shape[1],)),
    layers.Dense(32, activation='relu'),
    layers.Dense(16, activation='relu'),
    layers.Dense(1)
])
model_r.compile(optimizer='adam', loss='mse')
model_r.fit(Xr_tr_s, yr_tr, epochs=50, batch_size=32, validation_split=0.2, verbose=0)

p = model_r.predict(Xr_te_s).ravel()
res_r["TensorFlow"] = (mean_squared_error(yr_te, p) ** 0.5, r2_score(yr_te, p))
```
